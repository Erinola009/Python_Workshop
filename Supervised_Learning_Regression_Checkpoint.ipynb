```python
Energy_Data = Energy_Data.apply(pd.to_numeric, errors='coerce')
```

```python
# Checking Outliers
#Lets for example use IQR to dect outliers
Q1 = Energy_Data.quantile(0.25, axis=1)
Q3 = Energy_Data.quantile(0.75, axis=1)
IQR = Q3 - Q1
```

```python
dummies = pd.get_dummies(Energy_Data, drop_first=0.2)
dummies
```

```python
merged = pd.concat([Energy_Data, dummies], axis=1)
merged.drop(columns=["Time"], inplace=True)
merged.drop(columns=["BS"], inplace=True)
```

```python
merged
```

```python
# sorting the independent variable into x
y = merged.Energy
x = merged.drop("Energy", axis=1)
```

```python
x
```

```python
#Splitting Dataset into Training and Test Sets
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
# Import the Linear regression model
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
model.fit(x, y)
```

```python
#Training a Machine Learning Regression Algorithm
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
```

```python
#Assessing Model Performance Using Relevant Evaluation Metrics
from sklearn.metrics import mean_squared_error, r2_score
y_predict = model.predict(x_test)

MSE = mean_squared_error(y_test, y_predict)
r2 = r2_score(y_test, y_predict)
print(f'Mean Squared Error: {MSE}')
print(f'R² Score: {r2}')
```
